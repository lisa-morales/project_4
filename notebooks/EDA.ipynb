```python
import streamlit as st
import pandas as pd
import plotly_express as px
import numpy as np
 

df = pd.read_csv('../vehicles_us.csv') 
df['manufacturer'] = df['model'].apply(lambda x: x.split()[0])
```

```python
print(df.info())
(df.head())
```

```python
# Fill missing values in model_year with the median and convert to int
df['model_year'] = pd.to_numeric(df['model_year'], errors='coerce')
median_year = df['model_year'].median()
df['model_year'] = df['model_year'].fillna(median_year).astype(int)

# Convert cylinders to int with no decimal places
df['cylinders'] = pd.to_numeric(df['cylinders'], errors='coerce').fillna(0).astype(int)

#Fill in missing values in odometer with the median anc converting to int
df['odometer'] = pd.to_numeric(df['odometer'], errors='coerce')
median_odometer = df['odometer'].median()
df['odometer'] = df['odometer'].fillna(median_odometer).astype(int)

# Convert date_posted to datetime
df['date_posted'] = pd.to_datetime(df['date_posted'], errors='coerce')

```
