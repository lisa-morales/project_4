## Introduction
The purpose of this project is to use a dataset of listings for a wide variety of vehicles in various conditions. Data provided includes: price, model and model year, condition, cylinders, fuel type, mileage, vehicle type, paint color, is or is not 4WD, date posted, days listed and manufacturer. 

The dataframe is to be cleaned and analyzed for various factors of my choosing and displayed in appropriate charts with conclusions that can be drawn from the charts.

```python
import pandas as pd
import plotly.express as px
import numpy as np
import streamlit as st

df = pd.read_csv('../vehicles_us.csv') 
df['manufacturer'] = df['model'].apply(lambda x: x.split()[0])
```

```python
df.duplicated().sum()  # Check for duplicates
```

```python
print(df.info())
(df.head())
```

```python
########## MODEL_YEAR ########## 
# Replace NaN in model_year with the median model_year for each model
df['model_year'] = df['model_year'].fillna(
    df.groupby('model')['model_year'].transform('median')
)
# Fill any remaining NaN in model_year with overall median before converting to int
df['model_year'] = df['model_year'].fillna(df['model_year'].median())
df['model_year'] = pd.to_numeric(df['model_year'], errors='coerce').astype(int)


########## CYLINDERS ########## 
# Replace NaN/null values in cylinders with the median cylinders for each model
df['cylinders'] = df['cylinders'].fillna(
    df.groupby('model')['cylinders'].transform('median')
)
#Convert cylinders to int with no decimal places
df['cylinders'] = pd.to_numeric(df['cylinders'], errors='coerce').fillna(0).astype(int)


########## ODOMETER ########## 
# Replavce NaN/null values in odometer value 0
df['odometer'] = df['odometer'].fillna(0)
# Replace 0 values in odometer with the median based on model_year and model
odometer_median = df.groupby(['model_year'])['odometer'].transform('median')
df['odometer'] = df['odometer'].mask((df['odometer'].isna()) | (df['odometer'] == 0), odometer_median)
# Convert odometer to int with no decimal places
df['odometer'] = pd.to_numeric(df['odometer'], errors='coerce').fillna(0).astype(int)


########### PAINT_COLOR ###########
# Change NaN values in paint_color to 'unknown'
df['paint_color'] = df['paint_color'].fillna('unknown')


########### IS_4WD ##########
# Change NaN values in is_4wd to 0
df['is_4wd'] = df['is_4wd'].fillna(0).astype(int)
# Convert is_4wd to int
df['is_4wd'] = df['is_4wd'].astype(int)


########## DATE_POSTED ########## 
# Convert date_posted to datetime
df['date_posted'] = pd.to_datetime(df['date_posted'], errors='coerce')


########## DAYS_LISTED ##########
# Replace 0 values in days_listed with the median value
df.loc[df['days_listed'] == 0, 'days_listed'] = df['days_listed'].median()
```
